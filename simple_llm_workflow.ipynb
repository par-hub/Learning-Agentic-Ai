```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from dotenv import load_dotenv
from langchain_groq import ChatGroq

```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
class LLmState(TypedDict):
    Question: str
    answere: str
```

```python
def LLMQA(state: LLmState) -> LLmState:
    question = state['Question']
    llm = ChatGroq(model = "openai/gpt-oss-120b")
    prompt = f'Answere the following question{question}'
    answere = llm.invoke(prompt).content
    state['answere'] = answere
    return state
```

```python
graph = StateGraph(LLmState)
graph.add_node('llmQA',LLMQA)
graph.add_edge(START,'llmQA')
graph.add_edge('llmQA', END)
workflow = graph.compile()
```

```python
workflow.invoke({'Question': 'What is the capital city of India?'})
```
