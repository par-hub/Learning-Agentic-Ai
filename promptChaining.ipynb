```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file"
```

```python
llm = ChatGroq(model = "openai/gpt-oss-120b")
```

```python
class BlogState(TypedDict):
    topic: str
    outline: str
    blog:str
```

```python
def genoutline(state: BlogState)->BlogState:
    topic = state['topic']
    prompt = f'Generate an outine for the following topic{topic}'
    outline = llm.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def genoutblog(state: BlogState)->BlogState:
    topic = state['topic']
    outline = state['outline']
    prompt = f'Generate a blog for the following topic{topic}and outline{outline}'
    blog = llm.invoke(prompt).content
    state['blog'] = blog
    return state
```

```python
graph = StateGraph(BlogState)
graph.add_node('getoutline', genoutline)
graph.add_node('getblog',genoutblog)
graph.add_edge(START,'getoutline')
graph.add_edge('getoutline','getblog')
graph.add_edge('getblog',END)
workflow = graph.compile()
```

```python
finalstae = workflow.invoke({'topic': 'The impact of AI on society'})
```
