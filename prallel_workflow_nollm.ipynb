```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatmanState(TypedDict):
    runs: int
    balls : int
    fours:int
    sixes:int
    sr:float
    bpb:float
    b_p:float
    summary:str
```

```python
def claculatesr(state:BatmanState)->BatmanState:
    sr = (state['runs']/state['balls'])/100
    state[sr] = sr
    return {'sr':sr}
```

```python
def calculate_bpb(state: BatmanState):

    bpb = state['balls']/(state['fours'] + state['sixes'])

    return {'bpb': bpb}
```

```python
def calculate_boundary_percent(state: BatmanState):

    boundary_percent = (((state['fours'] * 4) + (state['sixes'] * 6))/state['runs'])*100

    return {'boundary_percent': boundary_percent}
```

```python
def summary(state: BatmanState):

    summary = f"""
Strike Rate - {state['sr']} \n
Balls per boundary - {state['bpb']} \n
Boundary percent - {state['boundary_percent']}
"""
    
    return {'summary': summary}
```

```python
graph = StateGraph(BatmanState)
graph.add_node('claculatesr',claculatesr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundary_percent',calculate_boundary_percent)
graph.add_node('summary',summary)
graph.add_edge(START,'claculatesr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_boundary_percent')
graph.add_edge('claculatesr','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_boundary_percent','summary')
graph.add_edge('summary',END)
workflow = graph.compile()
```

```python
intail_state = {
    'runs':100,
    'balls':50,
    'fours':10,
    'sixes':5

}
```

```python
workflow.invoke(intail_state)
```
